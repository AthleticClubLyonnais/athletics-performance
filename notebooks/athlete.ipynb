```python
# --- Construction ---
a1 = Athlete(licence="2275784", last_name="Perrin", first_name="Guillaume", yob=1982, sex="M")
a2 = Athlete(licence=9999999, last_name="Dupont", first_name=None, yob=2006, sex="F")  # int → str

print("=== repr ===")
print(a1)
print(a2)

print("\n=== fields ===")
print("licence    :", a1.licence)
print("last_name  :", a1.last_name)
print("first_name :", a1.first_name)
print("yob        :", a1.yob)
print("sex        :", a1.sex)

print("\n=== int licence coerced to str ===")
print(type(a2.licence), a2.licence)  # <class 'str'> 9999999

print("\n=== full_name (property) ===")
print(a1.full_name)   # "Guillaume Perrin"
print(a2.full_name)   # "Dupont" (no first name)

print("\n=== category(yos) ===")
print(a1.category(2026))   # "M1M"
print(a2.category(2026))   # "JUF"
a3 = Athlete(licence="3", last_name="X", first_name=None, yob=2012, sex="F")
print(a3.category(2026))   # "MIF"

print("\n=== sex validation ===")
try:
    Athlete(licence="0", last_name="Test", first_name=None, yob=2000, sex="X")
    print("ERROR: should have raised ValueError")
except ValueError as e:
    print(f"Correctly raised ValueError: {e}")

print("\n=== frozen (immutable) ===")
try:
    a1.last_name = "Martin"
    print("ERROR: should have raised FrozenInstanceError")
except Exception as e:
    print(f"Correctly raised {type(e).__name__}: {e}")

print("\n=== equality & hashing ===")
a1_copy = Athlete(licence="2275784", last_name="Perrin", first_name="Guillaume", yob=1982, sex="M")
print("a1 == a1_copy:", a1 == a1_copy)
print("a1 == a2     :", a1 == a2)
print("hash works   :", hash(a1) == hash(a1_copy))
```

```python
# --- Static method category_code(yob, yos[, sex]) ---
test_cases = [
    (1982, 2026, "M"),   # M1M
    (2006, 2026, "F"),   # ESF
    (2012, 2026, "F"),   # MIF
    (2010, 2026),        # CA  (no sex)
    (2008, 2026, "F"),   # JUF
    (2000, 2026, "M"),   # SEM
    (1960, 2026, "F"),   # M6F
    (2015, 2026, "M"),   # POM
    (2018, 2026, "F"),   # EAF
    (2021, 2026, "M"),   # BBM
]

print(f"{'yob':<6} {'yos':<6} {'sex':<5} {'category'}")
print("-" * 30)
for case in test_cases:
    yob, yos, *rest = case
    sex = rest[0] if rest else ""
    cat = Athlete.category_code(yob, yos, sex)
    print(f"{yob:<6} {yos:<6} {sex:<5} {cat}")
```
