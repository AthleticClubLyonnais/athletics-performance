```python
%load_ext autoreload
%autoreload 2
from athletics_performance import Club, DEPARTMENT_TO_LIGUE
```

```python
# --- Construction ---
c1 = Club(club_id="069106", name="ACL")
c2 = Club(club_id="075001", name="Paris AC")
c3 = Club(club_id="971001", name="Club Guadeloupe")

print("=== repr ===")
print(c1)
print(c2)
print(c3)
```

```python
# --- Fields (auto-derived from club_id) ---
print("=== fields ===")
print("club_id    :", c1.club_id)
print("name       :", c1.name)
print("dept_code  :", c1.dept_code)
print("ligue_code :", c1.ligue_code)
print("ligue_name :", c1.ligue_name)
```

```python
# --- Several clubs across ligues ---
sample_ids = [
    ("069106", "ACL"),
    ("075001", "Paris AC"),
    ("059001", "Lille AC"),
    ("013001", "Marseille AC"),
    ("067001", "Strasbourg AC"),
    ("033001", "Bordeaux AC"),
    ("031001", "Toulouse AC"),
    ("044001", "Nantes AC"),
    ("971001", "Club Guadeloupe"),
    ("974001", "Club Réunion"),
]

print(f"{'club_id':<10} {'dept':<6} {'code':<6} {'ligue'}")
print("-" * 55)
for club_id, name in sample_ids:
    c = Club(club_id=club_id, name=name)
    print(f"{c.club_id:<10} {c.dept_code:<6} {c.ligue_code:<6} {c.ligue_name}")
```

```python
# --- club_id length validation ---
print("=== club_id validation ===")
for bad_id in ["12345", "1234567", ""]:
    try:
        Club(club_id=bad_id, name="Test")
        print(f"{bad_id!r}: ERROR – should have raised ValueError")
    except ValueError as e:
        print(f"{bad_id!r}: correctly raised ValueError: {e}")
```

```python
# --- Unknown dept_code (no match in mapping) ---
c_unknown = Club(club_id="000001", name="Inconnu")
print("dept_code  :", c_unknown.dept_code)
print("ligue_code :", repr(c_unknown.ligue_code))   # ''
print("ligue_name :", repr(c_unknown.ligue_name))    # ''
```

```python
# --- frozen (immutable) ---
print("=== frozen (immutable) ===")
try:
    c1.name = "Changed"
    print("ERROR: should have raised FrozenInstanceError")
except Exception as e:
    print(f"Correctly raised {type(e).__name__}: {e}")
```

```python
# --- Equality & hashing ---
c1_copy = Club(club_id="069106", name="ACL")
print("=== equality & hashing ===")
print("c1 == c1_copy:", c1 == c1_copy)   # True
print("c1 == c2     :", c1 == c2)         # False
print("hash works   :", hash(c1) == hash(c1_copy))

print("\n=== usable as dict key ===")
rosters = {c1: ["Perrin", "Dupont"], c2: ["Martin"]}
print("c1 roster:", rosters[c1])
```

```python
# --- DEPARTMENT_TO_LIGUE lookup ---
print("=== DEPARTMENT_TO_LIGUE direct lookup ===")
for dept in ["069", "075", "971", "999"]:
    result = DEPARTMENT_TO_LIGUE.get(dept)
    print(f"{dept} → {result}")
```
